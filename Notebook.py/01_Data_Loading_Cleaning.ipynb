✅ Phase 2 — Data Cleaning
Objective

Prepare the dataset for analysis by handling missing values, standardizing columns, optimizing data types, and creating a clean version for SQL and Power BI.

```python
# Create a copy of the dataset
df_clean = df.copy()
```

```python
missing = df_clean.isnull().sum().sort_values(ascending=False)

missing[missing > 0]
```

```python
df_clean.dropna(inplace=True)

print(df_clean.shape)
```

```python
df_clean.isnull().sum().sum()
```

```python
df_clean.duplicated().sum()
```

```python
df_clean.rename(columns={
    "carrier":"Airline_Code",
    "carrier_name":"Airline",
    "airport":"Airport_Code",
    "airport_name":"Airport",
    "arr_flights":"Total_Flights",
    "arr_del15":"Delayed_Flights",
    "carrier_ct":"Carrier_Delay_Count",
    "weather_ct":"Weather_Delay_Count",
    "nas_ct":"NAS_Delay_Count",
    "security_ct":"Security_Delay_Count",
    "late_aircraft_ct":"Late_Aircraft_Delay_Count",
    "arr_cancelled":"Cancelled_Flights",
    "arr_diverted":"Diverted_Flights",
    "arr_delay":"Total_Delay_Minutes",
    "carrier_delay":"Carrier_Delay_Minutes",
    "weather_delay":"Weather_Delay_Minutes",
    "nas_delay":"NAS_Delay_Minutes",
    "security_delay":"Security_Delay_Minutes",
    "late_aircraft_delay":"Late_Aircraft_Delay_Minutes"
}, inplace=True)
```

```python
count_columns = [
    "Total_Flights",
    "Delayed_Flights",
    "Cancelled_Flights",
    "Diverted_Flights"
]

df_clean[count_columns] = df_clean[count_columns].astype("int64")
```

```python
numeric_columns = df_clean.select_dtypes(include=["number"]).columns

(df_clean[numeric_columns] < 0).sum()
```

```python
# Find rows with negative values
negative_rows = df_clean[df_clean["NAS_Delay_Minutes"] < 0]

negative_rows
```

```python
numeric_columns = df_clean.select_dtypes(include=["number"]).columns

for col in numeric_columns:
    count = (df_clean[col] < 0).sum()
    if count > 0:
        print(f"{col}: {count}")
```

```python
df_clean = df_clean[df_clean["NAS_Delay_Minutes"] >= 0]

print(df_clean.shape)
```

```python
(df_clean.select_dtypes(include="number") < 0).sum()
```

```python
df_clean.to_csv(
    "Cleaned_Airline_Delay.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
